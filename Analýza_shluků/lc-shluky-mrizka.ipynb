# Úloha: Mřížkové shlukování v podprostorech

Máte 3D tabulková data, kde reálné shluky existují pouze v 2D podprostoru ($X_1, X_2$), zatímco osa $X_3$ je tvořena čistým rovnoměrným šumem (uniform noise).

Metody jako K-Means nebo standardní DBSCAN v plném 3D prostoru selhávají nebo jsou degradovány přítomností neinformativní dimenze. Mřížkový přístup (CLIQUE) prostor diskretizuje a hledá husté jednotky (dense units) v jednotlivých podprostorech.

```python
import numpy as np

np.random.seed(42)
n_cluster = 400

# Shluky v podprostoru X1, X2 
c1 = np.random.normal(loc=[2.0, 2.0], scale=0.4, size=(n_cluster, 2))
c2 = np.random.normal(loc=[3.5, 3.5], scale=0.4, size=(n_cluster, 2))
subspace_data = np.vstack([c1, c2])

# Osa X3: šum 
# Var(X3) = 30^2 / 12 = 75.0 >> Var(X1, X2)
noise_dim = np.random.uniform(0.0, 30.0, size=(2 * n_cluster, 1))

X_grid_hard = np.hstack([subspace_data, noise_dim])
```
