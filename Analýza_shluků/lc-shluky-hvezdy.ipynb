# Úloha: Fyzikální vlastnosti hvězdných objektů

Dataset **star_data.csv** simuluje astrofyzikální měření (teplota $T$, zářivý výkon $L$, poloměr $R$) v logaritmických škálách.

## Úkoly

1. Načtěte soubor s daty do dataframu v pandas. Vizualizace pomocí pairplot.
2. Proveďte předzpracování pomocí standardního scaleru. Zdůvodněte, proč je škálování kritické pro algoritmy založené na eukleidovské vzdálenosti.
3. Hledání počtu shluků ($k$): Vypočtěte chybovost (parametr inertia) a průměrné skóre siluety (Silhouette score) pro $k \in \langle 2, 7 \rangle$. Vykreslete graf metody lokte (Elbow) a graf vývoje průměrné siluety. Určete optimální hodnotu $k$.
4. Pro nalezené optimální $k$ natrénujte: K-Means s parametry (n_init=10, random_state=42).
5. Natrénujte aglomerativní hierarchické shlukování (Agglomerative Clustering) s Wardovou vzdáleností (linkage='ward').
6. Porovnejte finální Silhouette skóre obou algoritmů. Vykreslete scatter ploty predikovaných shluků v rovině log_temp vs. log_lum pro oba modely.

### Nápověda
- Pro načtení: read_csv u pandas.
- Pro škálování: from sklearn.preprocessing import StandardScaler.
- Pro modely: from sklearn.cluster import KMeans, AgglomerativeClustering.
- Pro metriku: from sklearn.metrics import silhouette_score.
